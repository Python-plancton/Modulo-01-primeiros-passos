## Tópico 1 - Função len()

O tamanho de uma string pode ser obtido utilizando-se a função `len()`.
Essa função retorna um valor do tipo inteiro, representando a quantidade de caracteres contidos na string.

**Q1** - Crie uma variável chamada `nome` contendo seu nome.

Depois, imprima a quantidade de caracteres do nome usando `len()`

````python
len("O rato roeu a roupa do rei")
````

## Tópico 2 - Índice
Podemos acessar o conteúdo de uma string caractere a caractere, utilizando um número inteiro pra representar sua posição.
Esse número é chamado de `indice`, e começamos e contar de **zero**.

````python
alfabeto = "ABCDEFGH"
alfabeto[0]
````

Para acessar o indice ou posição de um caractere de uma string devemos colocar entre colchetes [ ].

````python
alfabeto = "ABCDEFGH"
alfabeto[3]
````

```

|    Índice |  0  |  1  |  2  |  3  |  4  |  5  |  6  |  7  |
| --------: | :-: | :-: | :-: | :-: | :-: | :-: | :-: | :-: |
| Caractere |  A  |  B  |  C  |  D  |  E  |  F  |  G  |  H  |


Assim, `alfabeto[3]` retorna o caractere **D**, pois a contagem dos índices em Python começa em **0**, e não em 1.

Observação: Indie e slicing não são a mesma coisa, como veremos mais adiante.
Os `índices` são usados para acessar partes específicas de um texto.
Em Engenharia e Análise de Dados, isso aparece o tempo todo durante a limpeza e transformação de dados.

## Tópico 3 - Concatenação
Para concatenar duas strings usamos o operador de adição `+`. 


````python
s = "ABC"
s= s + "C"
print(s)
````

Obs: As strings em Python são **imutáveis**. Isso significa que seu conteúdo não pode ser alterado diretamente.
Sempre que você faz uma concatenação, o Python cria uma nova string. Se quiser manter essa alteração,
é necessário atribuir o resultado a uma variável.

````python
nome = "Ana"
sobrenome = "Silva"

nome_completo = nome + " " + sobrenome

print(nome_completo)
````
