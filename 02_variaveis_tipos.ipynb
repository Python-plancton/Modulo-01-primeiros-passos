## Tópico 2 - Índice
Podemos acessar o conteúdo de uma string caractere a caractere, utilizando um número inteiro pra representar sua posição.
Esse número é chamado de `indice`, e começamos e contar de **zero**.

````python
alfabeto = "ABCDEFGH"
alfabeto[0]
````

Para acessar o indice ou posição de um caractere de uma string devemos colocar entre colchetes [ ].

````python
alfabeto = "ABCDEFGH"
alfabeto[3]
````

```

|    Índice |  0  |  1  |  2  |  3  |  4  |  5  |  6  |  7  |
| --------: | :-: | :-: | :-: | :-: | :-: | :-: | :-: | :-: |
| Caractere |  A  |  B  |  C  |  D  |  E  |  F  |  G  |  H  |


Assim, `alfabeto[3]` retorna o caractere **D**, pois a contagem dos índices em Python começa em **0**, e não em 1.

Observação: Indie e slicing não são a mesma coisa, como veremos mais adiante.
Os `índices` são usados para acessar partes específicas de um texto.
Em Engenharia e Análise de Dados, isso aparece o tempo todo durante a limpeza e transformação de dados.

## Tópico 3 - Concatenação
Para concatenar duas strings usamos o operador de adição `+`. 


````python
s = "ABC"
s= s + "C"
print(s)
````

Obs: As strings em Python são **imutáveis**. Isso significa que seu conteúdo não pode ser alterado diretamente.
Sempre que você faz uma concatenação, o Python cria uma nova string. Se quiser manter essa alteração,
é necessário atribuir o resultado a uma variável.

````python
nome = "Ana"
sobrenome = "Silva"

nome_completo = nome + " " + sobrenome

print(nome_completo)
````

As strings nome, " " (espaço) e sobrenome são concatenadas para formar uma nova string.

### Tópico 4 - Composição de strings (f-strings)

Embora seja possível juntar textos usando o operador +, em Python a forma mais recomendada é utilizar f-strings.

Uma f-string (formatted string) permite inserir o valor de variáveis diretamente dentro do texto, deixando o código mais simples e legível.

````python
nome = "Ana"
idade = 25

mensagem = "Meu nome é " + nome + " e tenho " + str(idade) + " anos."

print(mensagem)
````

Observe que foi necessário converter idade para texto usando `str()`, pois não é possível concatenar diretamente uma string com um `int`.

**Usando f-string**

````python
nome = "Ana"
idade = 25

mensagem = f"Meu nome é {nome} e tenho {idade} anos."

print(mensagem)
````

Nesse caso, o Python substitui automaticamente o conteúdo entre `{ }` 
pelo valor da variável correspondente, sem necessidade de converter números para texto.

💡 Boa prática

Hoje, as f-strings são a forma recomendada pela comunidade Python para combinar texto com variáveis. Elas deixam o código mais limpo, mais fácil de ler e evitam erros comuns de concatenação.

### Tópico 5 - Slicing
Até agora aprendemos que podemos acessar um único caractere de uma string usando um índice.

O índice representa a posição de cada caractere dentro da string. Em Python, a contagem sempre começa em 0.

````python
alfabeto = "ABCDEFGH"

print(alfabeto[0])  # A
print(alfabeto[3])  # D
print(alfabeto[7])  # H
````

| Índice |  0  |  1  |  2  |  3  |  4  |  5  |  6  |  7  |
| :----: | :-: | :-: | :-: | :-: | :-: | :-: | :-: | :-: |
|  Letra |  A  |  B  |  C  |  D  |  E  |  F  |  G  |  H  |


Também é possível utilizar índices negativos, que contam a partir do final da string.

| Índice positivo |  0  |  1  |  2  |  3  |  4  |  5  |  6  |  7  |
| :-------------: | :-: | :-: | :-: | :-: | :-: | :-: | :-: | :-: |
|      Letra      |  A  |  B  |  C  |  D  |  E  |  F  |  G  |  H  |
| Índice negativo |  -8 |  -7 |  -6 |  -5 |  -4 |  -3 |  -2 |  -1 |


````python
print(alfabeto[-1])  # H
print(alfabeto[-3])  # F
````

Os `índices` permitem acessar um único caractere da string.

Mas, muitas vezes, precisamos **extrair** apenas uma parte do texto.

Para isso, utilizamos o **slicing (fatiamento).**

**O que é Slicing?**

Slicing é uma técnica que permite extrair uma parte de uma string, criando uma nova string a partir da original.

A sintaxe é:
```python
string[inicio:fim:passo]
```

Onde:

- início → índice onde a extração começa.

- fim → índice onde a extração termina (esse índice não é incluído).

- passo → quantidade de posições que o Python avança (opcional).

````python
Exemplo 1 - Do início até um índice
````

````python
alfabeto = "ABCDEFGH"

print(alfabeto[:4])
````

Observe que o Python começa no início da string `[0]` e para antes do índice `4`.

Exemplo 2 - De um índice até o final

````python
alfabeto = "ABCDEFGH"

print(alfabeto[3:])
````

A extração começa no índice `3` e continua até o final da string.

Exemplo 3 - Entre dois índices

````python
alfabeto = "ABCDEFGH"

print(alfabeto[2:6])
````

Começa no índice `2` e termina antes do índice `6`.

Exemplo 4 - Utilizando passo

Também é possível definir de quantas em quantas posições o Python deve avançar.

````python
alfabeto = "ABCDEFGH"

print(alfabeto[::2])
````

Como o passo é `2`, o Python pega um caractere e pula o próximo.

Exemplo 5 - Invertendo uma string

Quando o passo é negativo, a leitura acontece da **direita para a esquerda**.

````python
alfabeto = "ABCDEFGH"

print(alfabeto[::-1])
````

Resumo:
| Expressão     | Significado                       |
| ------------- | --------------------------------- |
| `texto[3]`    | Caractere no índice 3             |
| `texto[:4]`   | Do início até antes do índice 4   |
| `texto[3:]`   | Do índice 3 até o final           |
| `texto[2:6]`  | Do índice 2 até antes do índice 6 |
| `texto[::2]`  | Um caractere a cada 2 posições    |
| `texto[::-1]` | String invertida                  |


Precisou extrair partes de códigos, datas, CPFs, CNPJs, telefones e outros identificadores durante processos de limpeza e transformação de dados? Via de Slicing!
